```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

Tokenize

```python
tokens = word_tokenize(document)
tokens
```

# Build a vocab

```python
vocab = {'<unk>' : 0 }
```

```python
for token in Counter(tokens).keys():
  if token not in vocab:
    vocab[token] = len(vocab)

vocab
```

```python
len(vocab)
```

# Extract senteces from existing data

```python
input_sentences = document.split("\n")
```

```python
input_sentences[4]
```

```python
def text_to_indices(sentence,vocab):
  numerical_sentences = []
  for token in sentence:
    if token in vocab:
      numerical_sentences.append(vocab[token])
    else:
      numerical_sentences.append(vocab['<unk>'])
  return numerical_sentences
```

```python
input_numeric_sentence = []

for sentence in input_sentences:
  input_numeric_sentence.append(text_to_indices(word_tokenize(sentence.lower()),vocab))
```

```python
print(input_numeric_sentence)
```

```python
training_sqeuence = []

for sentence in input_numeric_sentence:
  for i in range(1,len(sentence)):
    training_sqeuence.append(sentence[:i+1])
```

```python
training_sqeuence
```

```python
len(training_sqeuence)
```

```python
len_list = []
for sequence in training_sqeuence:
  len_list.append(len(sequence))
```

```python
max(len_list)
```

```python
padded_training_sequence = []
for sequence in training_sqeuence:
  padded_training_sequence.append([0] * (max(len_list)-len(sequence))+ sequence)
```

Convert into tensor

```python
padded_training_sequence = torch.tensor(padded_training_sequence,dtype=torch.long)
```

```python
X = padded_training_sequence[:, :-1]
```

Custom Dataset and DataLoader

```python
class CustomDataset(Dataset):
  def __init__(self,X,y):
    self.X = X
    self.y = y

  def __len__(self):
    return self.X.shape[0]

  def __getitem__(self, idx):
    return self.X[idx], self.y[idx]
```

```python
dataset = CustomDataset(X,y)
```

```python
len(dataset)
```

```python
dataloader = DataLoader(dataset,batch_size=32,shuffle=True)
```

```python
for input,output in dataloader:
  print(output)
```
