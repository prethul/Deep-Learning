Regular Expression

```python
import re
```

# selecting the number

```python
txt = "My name is Dipto howlader prethul, I am 23 years old"
x = re.findall("\d+",txt)
print(x)
```

# Selecting not the number

```python
txt = "My name is Dipto howlader prethul, I am 23 years old"
x = re.findall("\D+",txt)
print(x)
```

```python
txt = "I am dipto howlader prethul"
x = re.findall("\w+",txt)
print(x)
```

# match 2 digits

```python
txt = "There is 18 dollars"

x = re.findall("\d{2}",txt)
print(x)
```

```python
text = "there will be 234 23 1 students"

x = re.findall("\d{2,3}",text)
print(x)
```

```python
text = "This was my number 09182746463"

x = re.findall("\d{5}-\d{6}",text)
print(x)
```

# check at the begining number

```python
text = "123 45 it is a number"
x = re.findall("^\d{3}",text)
print(x)
```

# The last number selection

```python
text = "it was never 12-456-2027"
x = re.findall("\d+$",text)
print(x)
```

# wildcard character

```python
text = "Hello"
x = re.findall(r".",text)
print(x)
```

```python
text = "prethul@gmail.com"
x = re.findall(".......@.........",text)
print(x)
```
